# WAP to check which is greater number out of two numbers, input by user 

```python
x= int(input ("please enter your first number to check: "))
y= int(input ("please enter your second number to check: "))
if x>y:
    print(x,"is greater")

else:
     print(y, "is greater")

print ("program ends here!")
```

```python
x= int(input ("please enter your first number to check: "))
y= int(input ("please enter your second number to check: "))
z= int(input ("please enter your third number to check: "))
if x>y>z:
    print(x,"is greater")

elif y>x>z or y>z>x:
     print(y, "is greater")

else: 
    print(z ,"is greater")



print ("program ends here!")
```

```python
x = int(input("Enter 1st number: "))
y = int(input("Enter 2nd number: "))
z = int(input("Enter 3rd number: "))

if x > y and x > z:
    print(x,"is greater")
elif y > x and y > z:
    print(y,"is greater")
else:
    print(z,"is greater")
```

```python
#WAP to print string by entered by user. But it must skip all Vowels in it
#vowales a,e,i,o,u
Character=input("Enter the word of your choice: ")
for i in Character:
    if i == "a" or i == "e" or i == "i" or i == "o" or i == "u":
        continue

    else:
        print(i)
```

```python
Character=input("Enter the word of your choice:")
for i in Character:
    j= i.lower()
    if j == "a" or j == "e" or j == "i" or j == "o" or j == "u":
        continue

    else:
        print(i)
```

## While loop and For loop 

```python
Character=input("Enter the word of your choice:")
print(Character)
i=0
while i<len(Character):
    j= Character[i].lower()
    if j == "a" or j == "e" or j == "i" or j == "o" or j == "u":
        i=i+1
        continue

    else:
        print(Character[i])
        i=i+1
```

```python
def ADD(x,y):  # def Function_Name(Parameter)
    print(x+y)
def sub(x,y):
    print(x-y)
def mult(s,t):
    print(s*t)
def divison(d,o):
    print(d/o)
def add(x,y):
    return x+y


```

```python
ADD(30,40)
add(30,40)
```

# Local Variable & Global Vriable 

```python
def info():
    global x
    x=100
    print(x*400)

x=50
print(x)
info()
print(x*200)
```
