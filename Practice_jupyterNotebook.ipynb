# WAP to print whether a 3 digit number is an armstrong number

```python
def Armstrong():
    
   num =int(input("Enter any number to check whether Armstong or not: "))

   length = len(str(num))
   result=0
   for i in str(num):
    result = result + (int(i)**length)

   if result==num:
    print("this number",num,"is an Armstrong Number !")
   else:
    print("oh!, this number is  NOT an Armstrong Number.")
    
     
```

 # Factorial 

```python
def fact(n):
    if n==1:
        return 1
    else:
        product = n* fact(n-1)
        return(product)
fact(5)
```
